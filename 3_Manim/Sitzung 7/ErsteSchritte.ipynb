```python
%%manim -qm Shapes

class Shapes(Scene):
    def construct(self):
        circle = Circle()
        square = Square()
        triangle = Triangle()
    
        circle.shift(3*LEFT + 1*UP)
        square.shift(UP)
        triangle.next_to(circle, LEFT)
        triangle.shift(UP)
    
        self.add(circle, square, triangle)
```

Mögliche Mobjects sind u.A.
- Polygon,
- Rectangle,
- RegularPolygon,
- Square
- Star,
- Triangle,
- Line,
- Angle bzw. RightAngle,
- Dot,
- Tex bzw. MathTex

Der Koordinatenursprung wird mit `ORIGIN` bezeichnet. Die Richtungen sind `UP`, `DOWN`, `LEFT` und `RIGHT`.

Methoden wie `.shift()` können auch zusammengefügt werden. So haben folgende beiden Programmausschnitte die gleiche Funktion:

```python
square = Square()
square.shift(DOWN)

square = Square().shift(DOWN)
```

```python
%%manim -qm Shapes

class Shapes(Scene):
    def construct(self):
        circle = Circle()
        square = Square()
        triangle = Triangle()
    	
        circle.move_to(2 * LEFT)
        square.next_to(circle, LEFT)
        triangle.align_to(circle, RIGHT)
        self.add(circle, square, triangle)
```

```python
%%manim -qm MobjectStyling

class MobjectStyling(Scene):
    def construct(self):
        circle = Circle().shift(LEFT)
        square = Square().shift(UP)
        triangle = Triangle().shift(RIGHT)
    
        circle.set_stroke(color=GREEN, width=2)
        square.set_fill(color=YELLOW, opacity=0.1).set_stroke(color=YELLOW)
        triangle.set_fill(PINK, opacity=0.5)
    
        self.add(circle, square, triangle)
```

Die Reihenfolge der Mobjects kann man durch die Reihenfolge in `.add` ändern: `self.add(circle, square, triangle)` vs. `self.add(triangle, circle, square)`

```python
%%manim -qm SomeAnimations

class SomeAnimations(Scene):
    def construct(self):
        square = Square()
    
        self.play(FadeIn(square))

        self.wait()
    
        self.play(Rotate(square, PI/4))
    
        self.play(FadeOut(square))
    
        self.wait(1)
```

Wir können aber auch über die Methode `.animate` auf das Mobject zugreifen:

```python
%%manim -qm SomeAnimations

class SomeAnimations(Scene):
    def construct(self):
    	square = Square()
    	
    	self.add(square)
    	
    	self.play(square.animate.set_fill(WHITE, opacity=0.4))
    	
    	self.play(square.animate.shift(UP).rotate(PI/4), run_time = 3)
    	
    	self.play(FadeOut(square))
    	
    	self.wait(1)
```

```python
%%manim -qm SomeAnimations

class SomeAnimations(Scene):
    def construct(self):
    	tex = MathTex(r'\frac{10}{x}')
    	
    	self.add(tex)
    	
    	self.play(tex.animate.shift(UP).rotate(PI/4), run_time = 3)
    	
    	self.play(FadeOut(tex))
    	
    	self.wait(1)
```
