```python
from manim import *
```

```python
config.media_width = "50%"
config.verbosity = "WARNING"
```

```python
%%manim -qm DifferentWeight

class DifferentWeight(Scene):
    def construct(self):
        import manimpango

        g = VGroup()
        weight_list = dict(
            sorted(
                {
                    weight: manimpango.Weight(weight).value
                    for weight in manimpango.Weight
                }.items(),
                key=lambda x: x[1],
            )
        )
        for weight in weight_list:
            g += Text(weight.name, weight=weight.name, font="Sans")
        self.add(g.arrange(DOWN).scale(0.5))
```

```python
%%manim -qm DifferentWeight

class DifferentWeight(Scene):
    def construct(self):
        import manimpango
        text = Text("Test", weight=ULTRAHEAVY, font="Sans")
        self.add(text)
```

```python
%%manim -qm SimpleColor

class SimpleColor(Scene):
    def construct(self):
        col = Text("RED COLOR", color=RED)
        self.add(col)
```

```python
%%manim -qm GradientExample

class GradientExample(Scene):
    def construct(self):
        t = Text("Hello", gradient=(BLUE, RED, YELLOW), font_size=90)
        self.add(t)
```

```python
%%manim -qm MathTeXDemo

class MathTeXDemo(Scene):
    def construct(self):
        rtarrow0 = MathTex(r"\xrightarrow{x^6y^8}", font_size=96)
        rtarrow1 = Tex(r"$\xrightarrow{x^6y^8}$", font_size=96)

        self.add(VGroup(rtarrow0, rtarrow1).arrange(UP).to_edge(UP))
```

```python
%%manim -qm LaTeXSubstrings

class LaTeXSubstrings(Scene):
    def construct(self):
        tex = Tex('Hello', r'$\bigstar$', r'\LaTeX', font_size=144)
        tex.set_color_by_tex('igsta', RED)
        self.add(tex)
```
